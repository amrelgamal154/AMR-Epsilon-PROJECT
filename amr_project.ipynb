### Getting Insights from the data

Checking the highest sales per each country 

```python
df.groupby('Customer Country')['Sales per customer'].sum().sort_values(ascending=False).head(10).reset_index()
```

```python
df.groupby(['Order Country', 'Order City'])['Sales'].sum().sort_values(ascending=False).reset_index().round(10).head(30)
```

```python
df_us=df[df['Order Country']=='Estados Unidos']
plot=df_us.groupby('Order City')['Sales'].sum().sort_values(ascending=False).reset_index().round(2).head(20)
plot
```

```python
px.pie(data_frame=plot, values='Sales', names='Order City', title='Sales by City in the US')
```

```python
### Decision to Form a depot in the north east region of the US is based on the sales data. The top 15 cities in the US with the highest sales are: New York City
```

```python
northeast_total=df_us[df_us['Order City'].isin(['New York City','Philadelphia','Newark','Richmond'])]['Sales'].sum()

print(f"Total Sales for Northeast Cluster   : ${northeast_total:,.2f}")
```

```python
western_total=df_us[df_us['Order City'].isin(['Los Angeles','San Francisco','San Diego'])]['Sales'].sum()

print(f"Total Sales for Western Cluster     : ${western_total:,.2f}")

```

#### Check the sales trend over time 

```python
import datetime

datetime.datetime.today()
```

```python
df['order_date_formatted'] = pd.to_datetime(df['order date (DateOrders)'])
```

```python
df['order_date_formatted'].describe()
```

```python
df['order_year'] = df['order_date_formatted'].dt.year
df['order_year']
```

```python
df_sorted = df.sort_values(by= 'order_year')
df_sorted
```

```python
plot_df = df_sorted.groupby('order_year')['Sales'].sum().reset_index()
plot_df
```

```python
px.line(data_frame= plot_df, x= 'order_date_formatted', y= 'Sales')
```

```python
px.line(data_frame= plot_df, x= 'order_year', y= 'Sales',title='Sales trend over the year')
```
